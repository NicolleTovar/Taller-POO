Esta serie de diapositivas presenta una introducción estructurada al flujo de trabajo de Aprendizaje Automático (Machine Learning), abarcando herramientas, preprocesamiento, algoritmos y métricas de evaluación.
1. Ecosistema Tecnológico e Infraestructura
 * Herramientas de desarrollo: Se basa en el ecosistema de Python (Python Data Stack, Scikit-learn Suite).
 * Entorno y control de versiones: Uso de Google Colab en la nube interconectado con un servidor y Git / GitHub Enterprise para la gestión de código.
2. Flujo de Preprocesamiento de Datos
 * Imputación: Tratamiento y rellenado de datos nulos o faltantes.
 * Escalamiento: Normalización de escalas numéricas.
 * Reducción: Compresión dimensional de variables.
 * Clasificador: Estimador final para la predicción algorítmica.
3. Modelos y Algoritmos Principalmente Cubiertos
 * SVM y Truco del Kernel: Técnica que proyecta datos a dimensiones superiores para resolver problemas no lineales.
 * Ensemble Learning (Ensamble):
   * Bagging: Reduce la varianza entrenando árboles aleatorios en paralelo.
   * Boosting: Reduce el sesgo corrigiendo errores de forma iterativa.
 * Algoritmos No Supervisados:
   * K-Means: Agrupamiento utilizado para segmentación.
   * DBSCAN: Basado en densidad, aplicado a análisis demográficos.
   * Isolation Forest: Detección de anomalías aplicable a la prevención de fraudes.
4. Optimización, Evaluación y Errores Comunes
 * Búsqueda de Hiperparámetros:
   * GridSearchCV: Búsqueda exhaustiva sobre una malla (más precisa pero costosa en cómputo).
   * RandomizedSearchCV: Exploración automatizada y ágil mediante muestreo aleatorio.
 * Métricas Avanzadas de Evaluación: Matriz de confusión, Precisión, Exhaustividad (Recall), Puntuación F1-Score y Curvas ROC / Área Bajo la Curva (AUC).
 * Advertencia de Error Crítico: Se destaca la Fuga de Datos (Data Leakage), definida como la filtración inadvertida de información del conjunto de prueba hacia el conjunto de entrenamiento, lo cual invalida el modelo.
